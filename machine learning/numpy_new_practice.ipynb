```python
x = np.array([[952, 2, 1, 65],
              [1244, 3, 2, 64],
              [1947, 3, 2, 17],
              [1725, 3, 2, 42]])
mean = x.mean(axis=0)
sigma = x.std(axis=0)
z = (x - mean) / sigma
```

```python
x_new = np.array([1200, 3, 1, 40])

z_new = (x_new - mean) / sigma
# my guess for the shape before running it is 4, 
print(f"z: {z_new}, z_shape: {z_new.shape}")
```

```python
w = np.array([0.5, 0.2, 0.1, -0.3])
b = 350

prediction_row0 = np.dot(z[0], w) + b
print(f"Prediction: {prediction_row0}")
```

```python
prediction_all_four = np.dot(z, w) + b
print(prediction_all_four)
```
