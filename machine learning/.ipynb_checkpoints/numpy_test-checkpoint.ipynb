```python
errors = np.random.randint(1, 41, (40))
x = np.random.randint(1, 100, (40, 3))

weighted_sum = errors @ x
print(weighted_sum)

new_weighted_sum = np.zeros(3)

for i in range(0, 3):
    new_weighted_sum[i] = (errors * x[:, i]).sum()

print(new_weighted_sum)
```
