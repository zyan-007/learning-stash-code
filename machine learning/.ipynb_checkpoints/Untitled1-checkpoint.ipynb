```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x_train1 = np.array([0., 1, 2, 3, 4, 5])
y_train1 = np.array([0, 0, 0, 1, 1, 1])
x_train2 = np.array([[0.5, 1, 1.5, 3, 2, 1], [1.5, 1, 0.5, 0.5, 2, 2.5]])
x_train2 = x_train2.T
y_train2 = np.array([0,0,0,1,1,1])
```

```python
pos = y_train1 == 1
neg = y_train1 == 0

fig, ax = plt.subplots(1, 2,figsize=(8,3))

ax[0].scatter(x_train1[pos], y_train1[pos], marker='x', c='r', label="y=1", lw=3, s=80)
ax[0].scatter(x_train1[neg], y_train1[neg], marker='o', facecolor='none', edgecolor='b', lw=3, s= 100, label="y=0")
ax[0].set_ylabel('y', fontsize=12)
ax[0].set_xlabel('x', fontsize=12)
ax[0].set_title('one variable plot')
ax[0].legend()

pos = y_train2 == 1
neg = y_train2 == 0

ax[1].scatter(x_train2[pos, 0], x_train2[pos, 1], marker='x', c='r' ,label='y=1', lw=3, s=100)
ax[1].scatter(x_train2[neg, 0], x_train2[neg, 1], marker='o', edgecolor='b', facecolor='none', label='y=2', lw=3, s=80)
ax[1].set_ylabel(r'$x_1$', fontsize=12)
ax[1].set_xlabel('$x_0$', fontsize=12)
ax[1].set_title('two variable plot')

plt.tight_layout()
plt.show()
```
