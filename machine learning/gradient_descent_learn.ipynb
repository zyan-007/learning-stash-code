```python
import math
import copy
import matplotlib.pyplot as plt
import numpy as np
```

```python
x_train = np.array([1.0, 2.0])
y_train = np.array([300.0, 500.0])
```

```python
def compute_cost(x, y, w, b):
    cost = 0
    m = len(x)
    for i in range(m):
        cost += (((x[i] * w) +b) - y[i]) ** 2
    cost /= (2*m)
    return cost
```

```python
def compute_gradient(x, y, w, b):

    m = len(x)
    dj_dw = 0
    dj_db = 0
    for i in range(m):
        dj_dw += ((x[i] * w + b) - y[i]) * (x[i]) 
        dj_db += ((x[i] * w + b) - y[i])
    dj_dw /= m
    dj_db /= m
    return dj_dw, dj_db
```

```python
def gradient_descent(x, y, w_int, b_int, alpha, iterations):

    w = w_int
    b = b_int
    for i in range(iterations):
        dj_dw, dj_db = compute_gradient(x, y, w, b)

        temp_w = w - (alpha * dj_dw)
        temp_b = b - (alpha * dj_db)

        w = temp_w
        b = temp_b
        
    
    return w, b
```

```python
w = 0
b = 0
iterations = 10000
alpha = 1.0e-2

final_w, final_b = gradient_descent(x_train, y_train, w, b, alpha, iterations)

print(f"final values are w: {final_w}, b: {final_b}")
```
