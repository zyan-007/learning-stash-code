```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x_train = np.array([1.0, 1.5, 2.0, 2.5, 3.0, 3.5, 4.0])
y_train = np.array([210, 265, 360, 415, 505, 590, 640])

w_values = [100, 125, 150, 175, 200]
b_values = [0, 25, 50, 75, 100]
m = len(x_train)
```

```python
# showing data on graph
plt.scatter(x_train, y_train, marker='o', c='r')
plt.title("House price vs Size")
plt.xlabel("Size (1000 sqft)")
plt.ylabel("Price (1000s of dollar)")
plt.show()
```

```python
def compute_linear_regression(x_train, w, b):
    prediction_y = np.zeros(m);
    for i in range(m):
        prediction_y[i] = x_train[i] * w + b
    return prediction_y
```

```python
def compute_cost_function(y_train, predictions):
    loss = 0
    for i in range(m):
        loss += (y_train[i] - predictions[i]) ** 2;
    loss /= 2 * m
    return loss
        
```

```python
loss = np.zeros(len(w_values))
for i in range(len(w_values)):
    prediction = compute_linear_regression(x_train, w_values[i], b_values[i]);
    
    #plotting each graph as well
    plt.title(f"Graph {i+1}, w = {w_values[i]}, b = {b_values[i]}")
    plt.xlabel("Size(1000 sqft)")
    plt.ylabel("Price(1000s of dollar)")
    plt.scatter(x_train, y_train, marker='o', c='r')
    plt.plot(x_train, prediction, c='b', label=f'prediction{i+1}')
    plt.legend()
    plt.show()
    print()
        
    loss[i] = compute_cost_function(y_train, prediction)
```

```python
minimum_cost = min(loss)
index = int(np.where(loss == minimum_cost)[0][0])
print(index)
print(f"Best values for parameters w and b are {w_values[index]} and {b_values[index]}, with a cost function value of {minimum_cost}")
```

```python
# graph for cost vs w function 
plt.title("Graph")
plt.xlabel("w")
plt.ylabel("Cost")
plt.plot(w_values, loss, c='b')
plt.show()
```

```python
# graph for cost vs b function 
plt.title("Graph")
plt.xlabel("b")
plt.ylabel("Cost")
plt.plot(b_values, loss, c='b')
plt.show()
```
